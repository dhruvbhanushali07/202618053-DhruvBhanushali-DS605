```python
M1 = np.array([
    [1, 2],
    [3, 4]
])

M2 = np.array([
    [5, 6],
    [7, 8]
])

```

```python
addition = M1 + M2

print(addition)

```

```python

matrix_product = M1 @ M2

print(matrix_product)

```

```python

transpose = M1.T

print(transpose)
```

```python
det = np.linalg.det(M1)

print("Determinant:", det)
```

```python
if not np.isclose(np.linalg.det(M1), 0):
    inverse = np.linalg.inv(M1)
    print("Inverse:")
    print(inverse)
```

```python
identity = M1 @ inverse

print(identity)
print(np.allclose(identity, np.eye(M1.shape[0])))
```

```python
np.random.seed(42)

data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

print(data)

```

```python
sample_mean = np.mean(data)

print("Sample Mean:", sample_mean)
```

```python
sample_std = np.std(data, ddof=1)

print("Sample Standard Deviation:", sample_std)
```

```python
print("Chosen Mean:", 50)
print("Sample Mean:", sample_mean)

print("Chosen Standard Deviation:", 10)
print("Sample Standard Deviation:", sample_std)
```

```python
import matplotlib.pyplot as plt

plt.hist(data, bins=30, edgecolor="black")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")

plt.show()
```
