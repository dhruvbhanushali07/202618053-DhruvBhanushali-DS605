```python

matrix_product = M1 @ M2

print(matrix_product)

```

```python

transpose = M1.T

print(transpose)
```

```python
det = np.linalg.det(M1)

print("Determinant:", det)
```

```python
if not np.isclose(np.linalg.det(M1), 0):
    inverse = np.linalg.inv(M1)
    print("Inverse:")
    print(inverse)
```

```python
identity = M1 @ inverse

print(identity)
print(np.allclose(identity, np.eye(M1.shape[0])))
```

Task 5 - Filtering and Querying
Use Boolean indexing and/or query() to answer:
+ Male passengers older than 50: how many?
+ Female first-class passengers: how many, and what percentage survived?
+ Age 20-40, Fare above overall median, and survived: how many?
+ Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
+ Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python

males_over_50 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
print(f"Male passengers older than 50: {len(males_over_50)}")

female_pclass1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
count_f_pclass1 = len(female_pclass1)
survived_pct_f_pclass1 = (female_pclass1['Survived'].mean()) * 100
print(f"Female 1st-class passengers: {count_f_pclass1}")
print(f"Survival percentage: {survived_pct_f_pclass1:.2f}%")

overall_median_fare = df['Fare'].median()
q3 = df[(df['Age'] >= 20) & (df['Age'] <= 40) & (df['Fare'] > overall_median_fare) & (df['Survived'] == 1)]
print(f"Age 20-40, Fare > median ({overall_median_fare:.2f}), and survived: {len(q3)}")

q4 = df[(df['SibSp'] == 0) & (df['Parch'] == 0) & (df['Age'] < 30) & (df['Survived'] == 0)]
print(f"Traveling alone, age < 30, and did not survive: {len(q4)}")

southampton_median_fare = df[df['Embarked'] == 'S']['Fare'].median()
q5 = df.query("Embarked == 'S' and Pclass in [2, 3] and Fare > @southampton_median_fare")
print(f"Embarked 'S', Pclass 2 or 3, Fare > Southampton median ({southampton_median_fare:.2f}): {len(q5)}")
```

Task 6 - groupby() and Aggregation
+ Survival rate by Sex.
+ Survival rate by Pclass.
+ Average Age and Fare by Pclass.
+ Passenger count and survival rate by Sex-Pclass.
+ Passenger count, average Fare, and survival rate by Embarked.

```python
print("--- Survival Rate by Sex ---")
display(df.groupby('Sex')['Survived'].mean().reset_index())

print("\n--- Survival Rate by Pclass ---")
display(df.groupby('Pclass')['Survived'].mean().reset_index())

print("\n--- Average Age and Fare by Pclass ---")
display(df.groupby('Pclass')[['Age', 'Fare']].mean().reset_index())

print("\n--- Count and Survival Rate by Sex and Pclass ---")
display(df.groupby(['Sex', 'Pclass'])['Survived'].agg(['count', 'mean']).reset_index().rename(columns={'mean': 'Survival Rate', 'count': 'Passenger Count'}))

print("\n--- Stats by Embarked Port ---")
display(df.groupby('Embarked').agg(
    Passenger_Count=('PassengerId', 'count'),
    Average_Fare=('Fare', 'mean'),
    Survival_Rate=('Survived', 'mean')
).reset_index())
```

Task 7 - Missing Values and Fare Outliers
+ Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.
+ Fill missing Age values with the mean Age. Show missing counts before and after.
+ Try to fill NaN values by imputing using: mean, median, mode, random value
+ For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.

```python
missing_df = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage (%)': (df.isnull().sum() / len(df)) * 100
})
print("--- Missing Values Summary ---")
display(missing_df[missing_df['Missing Count'] > 0])

plt.figure(figsize=(8, 4))
df.isnull().sum()[df.isnull().sum() > 0].plot(kind='bar', color='coral')
plt.title('Missing Value Counts Per Column')
plt.ylabel('Count')
plt.xlabel('Columns')
plt.xticks(rotation=0)
plt.show()

print("Missing Age before imputation:", df['Age'].isnull().sum())

df_imputed = df.copy()

df_imputed['Age_Mean'] = df_imputed['Age'].fillna(df_imputed['Age'].mean())

df_imputed['Age_Median'] = df_imputed['Age'].fillna(df_imputed['Age'].median())

df_imputed['Age_Mode'] = df_imputed['Age'].fillna(df_imputed['Age'].mode()[0])

random_ages = np.random.choice(df['Age'].dropna(), size=df['Age'].isnull().sum())
df_imputed['Age_Random'] = df_imputed['Age']
df_imputed.loc[df_imputed['Age_Random'].isnull(), 'Age_Random'] = random_ages

df['Age'] = df['Age'].fillna(df['Age'].mean())
print("Missing Age after mean imputation:", df['Age'].isnull().sum())

Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]

print(f"\n--- Fare Outliers (IQR Method) ---")
print(f"Q1: {Q1:.2f} | Q3: {Q3:.2f} | IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f} | Upper Bound: {upper_bound:.2f}")
print(f"Number of Fare Outliers: {len(outliers)}")
```

Task 8 - Features and Pivot Table
+ Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
+ Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest
groups.

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

print("First 5 rows with new features:")
display(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head())

pivot_survived = df.pivot_table(index='Sex', columns='Pclass', values='Survived', aggfunc='mean')
print("\n--- Pivot Table: Survival Rate by Sex and Pclass ---")
display(pivot_survived)

stacked = pivot_survived.unstack()
highest_group = stacked.idxmax()
highest_val = stacked.max()
lowest_group = stacked.idxmin()
lowest_val = stacked.min()

print(f"Highest Survival Group: Pclass {highest_group[0]}, Sex '{highest_group[1]}' ({highest_val * 100:.2f}%)")
print(f"Lowest Survival Group:  Pclass {lowest_group[0]}, Sex '{lowest_group[1]}' ({lowest_val * 100:.2f}%)")
```

## Observation

Strong Gender Disparity in Survival: Female passengers had a drastically higher survival rate (~74.2%) compared to male passengers (~18.9%), reflecting the "women and children first" evacuation protocol.

Socioeconomic Class Influence (Pclass): Passenger class shows a strong negative correlation with survival. 1st class had the highest survival rate (~63%), while 3rd class had the lowest (~24%).

Combined Socioeconomic and Gender Bias: First-class female passengers had the highest overall survival rate (>96%), whereas 3rd-class male passengers had the lowest survival rate (<14%).

Impact of Ticket Fare: Fare exhibits a positive correlation with survival. Passengers paying higher fares (especially those above the upper IQR bound of ~$66.30) were much more likely to survive, primarily because high fares correspond to 1st class tickets.

Port of Embarkation Differences: Passengers embarking from Cherbourg (Embarked = 'C') had the highest average fare and the highest survival rate (~55%) compared to those from Southampton ('S') and Queenstown ('Q').

Missing Data Concentration: The Cabin column is heavily missing (~77% missing data), while Age has ~20% missing values, making Age a candidate for mean/median imputation while Cabin requires special handling or dropping.
